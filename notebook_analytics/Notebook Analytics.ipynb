## 5 Perguntas Analyticas

Objetivo: Validar o modelo e demostrando o consumo analiticos do Pipeline criando via Azure Databrikcs

## 1️⃣ Qual o nível de experiência mais comum na base?

```sql
SELECT  
  nivel_experiencia,
  COUNT(*) AS Qtde_Total
FROM `cat-adls-br-dev`.gold.ti_salaries
GROUP BY nivel_experiencia
ORDER BY Qtde_Total DESC;
```

###  2️⃣ Top 10 países com maior média salarial

```sql
SELECT 
    pais_residencia,
    COUNT(*) AS Qtde
FROM `cat-adls-br-dev`.silver.ti_salaries
GROUP BY pais_residencia
ORDER BY Qtde DESC
LIMIT  10;
```

### 3️⃣ Existe relação entre trabalho remoto e salário?

```sql
SELECT 
    nivel_experiencia,
    percentual_remoto,
    ROUND(
        (COUNT(*) * 100.0) /
        SUM(COUNT(*)) OVER (PARTITION BY nivel_experiencia), 2) AS Percentual
FROM `cat-adls-br-dev`.silver.ti_salaries
GROUP BY nivel_experiencia, 
percentual_remoto
ORDER BY 
nivel_experiencia,
Percentual DESC;
```

## 4️⃣ Qual o percentual de profissionais remotos vs presenciais?

```sql
SELECT
    percentual_remoto,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentual
FROM `cat-adls-br-dev`.gold.ti_salaries
GROUP BY percentual_remoto
ORDER BY percentual DESC;
```

## 5️⃣ Qual o impacto da senioridade no trabalho remoto?

```sql
SELECT
    nivel_experiencia,
    percentual_remoto,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (PARTITION BY nivel_experiencia),
        2
    ) AS percentual
FROM `cat-adls-br-dev`.gold.ti_salaries
GROUP BY
    nivel_experiencia,
    percentual_remoto
ORDER BY
    nivel_experiencia,
    percentual DESC;
```
