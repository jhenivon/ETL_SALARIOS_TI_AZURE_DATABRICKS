## Observação
A camada Bronze é utilizada para leitura e Data Profiling dos dados. A camada Silver foi construída a partir dos dados da Bronze, porém não foi persistida na Bronze. Optamos por realizar as transformações utilizando CTEs, que não persistem os dados intermediários. Dessa forma, as transformações da Silver são executadas em tempo de execução a partir da Bronze, e o resultado final é persistido na camada Silver.
