```python
# inicializa o SparkSesion
from pyspark.sql import SparkSession
import pyspark.sql.functions as F
from pyspark.sql.functions import when, col, count, avg, round
```

```python
df_raw = spark.read.csv("abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/raw/ti_salaries.csv", header=True, inferSchema=True)   
df_raw.printSchema()  
df_raw.show(10)                            
```

## Salvando df_raw na Camada Bronze

```python
df_raw.write \
    .format("delta") \
    .mode("overwrite") \
    .option(
        "path",
        "abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/bronze/ti_salaries"
    ) \
    .saveAsTable("`cat-adls-br-dev`.bronze.ti_salaries")
```
